```python
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")

DATABASE_URL = (
    f"postgresql://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(DATABASE_URL)
```

## Топ треки

```python
with open("../sql/analytics/top_tracks.sql", "r") as file:
    top_tracks_query = file.read()

top_tracks_df = pd.read_sql(top_tracks_query, engine)
```

```python
top_tracks_df.to_csv(
    "../data/exports/top_tracks.csv",
    index=False
)
```

## Дневные метрики

```python
with open("../sql/analytics/daily_metrics.sql", "r") as file:
    daily_metrics_query = file.read()

daily_metrics_df = pd.read_sql(daily_metrics_query, engine)
```

```python
daily_metrics_df.to_csv(
    "../data/exports/daily_metrics.csv",
    index=False
)
```
